# 00_get_data: unduh dataset ROBOD

Pengganti `scripts/00_get_data.sh`. Mengambil 5 file `combined_Room<k>.csv` (Tekler et al., 2022) ke `data/raw`. Bila salinan sudah ada di `../data/raw` (folder induk), file disalin dari sana; jika tidak, repo `ideas-lab-nus/robod` di-clone dari GitHub.

```python
import shutil
import subprocess
import tempfile
from pathlib import Path

ROOT = Path.cwd().resolve().parent
RAW = ROOT / "data" / "raw"
RAW.mkdir(parents=True, exist_ok=True)
local = ROOT.parent / "data" / "raw"

if (RAW / "combined_Room3.csv").exists():
    print("data already present")
elif (local / "combined_Room3.csv").exists():
    for f in sorted(local.glob("combined_Room*.csv")):
        shutil.copy2(f, RAW / f.name)
    print("copied from", local)
else:
    with tempfile.TemporaryDirectory() as tmp:
        subprocess.run(["git", "clone", "--depth", "1", "https://github.com/ideas-lab-nus/robod", f"{tmp}/robod"],
                       check=True)
        for f in sorted(Path(tmp, "robod", "Data").glob("combined_Room*.csv")):
            shutil.copy2(f, RAW / f.name)
for f in sorted(RAW.glob("*.csv")):
    print(f"{f.name:24s} {f.stat().st_size / 1e6:6.1f} MB")
```
