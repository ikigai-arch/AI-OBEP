# 00_get_data: unduh dataset ROBOD

Pengganti `scripts/00_get_data.sh`. Selalu meng-clone repo publik `ideas-lab-nus/robod` dari GitHub dan menyalin 5 file `combined_Room<k>.csv` ke `data/raw` (menimpa salinan lama).

Catatan: salinan lama di `../data/raw` (file Juli 2022) **tidak dipakai**. Energi chilled water di versi itu jauh lebih kecil, sehingga hasilnya tidak sesuai laporan (R² surrogate ≈ 0,66, bukan 0,94).

```python
import shutil
import subprocess
import tempfile
from pathlib import Path

ROOT = Path.cwd().resolve().parent
RAW = ROOT / "data" / "raw"
RAW.mkdir(parents=True, exist_ok=True)

with tempfile.TemporaryDirectory() as tmp:
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/ideas-lab-nus/robod", f"{tmp}/robod"],
                   check=True)
    commit = subprocess.run(["git", "-C", f"{tmp}/robod", "log", "-1", "--format=%h %ad", "--date=short"],
                            capture_output=True, text=True).stdout.strip()
    for f in sorted(Path(tmp, "robod", "Data").glob("combined_Room*.csv")):
        shutil.copy2(f, RAW / f.name)
print("ROBOD commit:", commit)
for f in sorted(RAW.glob("*.csv")):
    print(f"{f.name:24s} {f.stat().st_size / 1e6:6.1f} MB")
```
